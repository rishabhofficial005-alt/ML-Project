```python
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import FunctionTransformer
from sklearn.preprocessing import OrdinalEncoder
```

```python
skewed_col = ['Study_Hours']

other_numeric_cols = [
"Age",
"Avg_Daily_Usage_Hours",
"Daily_Unlocks",
"Physical_Activity_Hours",
"Sleep_Hours_Per_Night"
]

ordinal_col = ['Stress_Level']

normal_col = [
"Gender",
"Academic_Level",
"Most_Used_Platform",
"Purpose_Of_Use",
"Grouped_Country"
]
```

```python
skew_pipeline=Pipeline(
    steps=[
        ('log_transformer',FunctionTransformer(np.log1p)),
        ('scaler',StandardScaler())
    ]
)

plain_pipeline=Pipeline(
    steps=[
        ('scaler',StandardScaler())
    ]
)
ordinal_pipeline=Pipeline(
    steps=[
        ('ordinal_encoder',OrdinalEncoder(categories=[order]))
    ]
)
normal_pipeline=Pipeline(
    steps=[
        ('one_hot_encoder',OneHotEncoder(handle_unknown='ignore'))
    ]

)
```

```python
preprocessor=ColumnTransformer(
    [
        ('skew_pipeline',skew_pipeline,skewed_col),
        ('plain_pipeline',plain_pipeline,other_numeric_cols),
        ('ordinal_col',ordinal_pipeline,ordinal_col),
        ('normal_col',normal_pipeline,normal_col)
    ]
)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,r2_score
lr_pipeline=Pipeline(
    steps=[
        ('preprocessor',preprocessor),
        ('model',LinearRegression())
    ]
)
```

```python
lr_pipeline.fit(X_train,y_train)
```

```python
y_pred=lr_pipeline.predict(X_test)
```

```python
from sklearn.metrics import mean_squared_error,r2_score
mse=mean_squared_error(y_test,y_pred)
rmse=np.sqrt(mse)
r2=r2_score(y_test,y_pred)
print(f"The mean squared error: {mse}")
print(f"The root mean squared error: {rmse}")
print(f"The r2 score: {r2}")
```

```python
from sklearn.ensemble import RandomForestRegressor
rf_pipeline=Pipeline(
    steps=[
        ('preprocessor',preprocessor),
        ('model',RandomForestRegressor(random_state=42))
    ]
)
```

```python
rf_pipeline.fit(X_train,y_train)
```

```python
y_pred_rand=rf_pipeline.predict(X_test)
```

```python
mse_random=mean_squared_error(y_test,y_pred_rand)
rmse_random=np.sqrt(mse_random)
r2_random=r2_score(y_test,y_pred_rand)
print(f"The mean squared error: {mse_random}")
print(f"The root mean squared error: {rmse_random}")
print(f"The r2 score: {r2_random}")
```

**Hyperparameter Tuning**

```python
from sklearn.model_selection import RandomizedSearchCV
param_grid = {
    'model__n_estimators'       : [100,200,300],
    'model__max_depth'         : [5,10,15],
    'model__min_samples_split' : [2,5,10],
    'model__min_samples_leaf'  : [1,2,4]
}

random_search = RandomizedSearchCV(
    estimator=rf_pipeline,
    param_distributions=param_grid,
    n_iter = 15,
    cv = 5,
    scoring='r2',
    random_state=42,
    n_jobs =-1 #processor free
)

random_search.fit(X_train, y_train)
```

```python
random_search.best_params_
```

```python
rf_final_pipeline=Pipeline(
    steps=[
        ('preprocessor',preprocessor),
        ('model',RandomForestRegressor(n_estimators=200,min_samples_split=5,min_samples_leaf=2,max_depth=15,random_state=42))
    ])
```

```python
rf_final_pipeline.fit(X_train,y_train)
```

```python
y_pred_final=rf_final_pipeline.predict(X_test)
```

```python
r2_random=r2_score(y_test,y_pred_final)
print(f"The r2 score: {r2_random}")
#
```
